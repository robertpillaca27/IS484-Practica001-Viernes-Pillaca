# Ejercicio 3: Inspección Visual de un Producto

### Ficha PEAS
* **Medida de desempeño (P):** Porcentaje de artículos defectuosos descartados correctamente, tasa de falsos desechos y costo de reprocesamiento.
* **Entorno (E):** Cinta transportadora de la línea de producción e iluminación industrial.
* **Acciones (A):** `Aprobar`, `Revisión manual`, `Rechazar`.
* **Percepción (S):** Matriz de $5 \times 5$ píxeles dada por la cámara ($0 = \text{normal}, 1 = \text{defectuoso}$).

### Justificación de las reglas
Un producto con 0 píxeles defectuosos cumple los estándares de calidad de fábrica para ser aprobado. Un rango de 1 a 3 defectos indica imperfecciones leves que pueden ser corregidas por un operario o confirmadas como falsos positivos de la cámara. Más de 3 píxeles defectuosos denota una falla estructural grave en la pieza cuyo costo de reparación supera el costo de descarte.

```python
### Código
import numpy as np

def agente_inspeccion(imagen):
    """
    Percepcion (S): matriz 5x5 de NumPy (0 = normal, 1 = defectuoso)
    """
    defectos = np.sum(imagen == 1)
    
    if defectos == 0:
        return "aprobar", "producto sin defectos detectados"
    elif defectos <= 3:
        return "revision manual", f"se detectaron {defectos} pixeles defectuosos"
    else:
        return "rechazar", f"se detectaron {defectos} pixeles defectuosos (supera el limite)"
```

# Ejercicio 2: Escalamiento de Tickets de Soporte Técnico

### Ficha PEAS
* **Medida de desempeño (P):** Tiempo promedio de resolución de tickets, tasa de retención de clientes VIP y número de tickets mal derivados.
* **Entorno (E):** Sistema de mesa de ayuda (Helpdesk) y cola de tickets de soporte técnico.
* **Acciones (A):** `Asignar a Nivel 1`, `Escalar a Nivel 2`, `Escalar a Nivel 3`.
* **Percepción (S):** `tiempo_espera_minutos`, `nivel_urgencia`, `cliente_premium`.

### Justificación de las reglas
La urgencia alta compromete directamente la operación del usuario, exigiendo atención inmediata por especialistas (Nivel 3). Un cliente premium representa mayor valor de negocio, por lo que si espera más de 15 minutos debe recibir soporte crítico prioritario. Los tiempos de espera prolongados (>30 min) en tickets comunes denotan cuellos de botella que justifican elevar la prioridad a Nivel 2 para garantizar la satisfacción del cliente.

```python
### Código
def agente_soporte(tiempo_espera_minutos, nivel_urgencia, cliente_premium):
    urg = nivel_urgencia.lower()
    
    if urg == "alta" or (cliente_premium and tiempo_espera_minutos > 15):
        motivo = "urgencia alta reportada" if urg == "alta" else "cliente premium con tiempo de espera critico (>15 min)"
        return "escalar a nivel 3", motivo
    elif urg == "media" or cliente_premium or tiempo_espera_minutos > 30:
        if urg == "media":
            motivo = "urgencia media requiere atencion especializada"
        elif cliente_premium:
            motivo = "atencion prioritaria por estatus de cliente premium"
        else:
            motivo = "tiempo de espera supero el limite tolerable de 30 minutos"
        return "escalar a nivel 2", motivo
    else:
        return "asignar a nivel 1", "urgencia baja dentro del tiempo de espera estandar"
```
